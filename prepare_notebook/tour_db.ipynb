```python
pet_df = pd.read_csv("./data/pet_tour_data.csv")
```

```python
# 1. 삭제할 컬럼 리스트 정의 
cols_to_drop = [
    'addr2', 'areacode', 'cat1', 'cat2', 'cat3', 'contentid', 
 'createdtime', 'firstimage', 'tel', 
    'showflag', 'lDongRegnCd', 'lDongSignguCd', 
    'lclsSystm1', 'lclsSystm2', 'lclsSystm3', 'firstimage2', 'cpyrhtDivCd', 'mlevel', 'modifiedtime', 'sigungucode' 
]

# 2. 컬럼 삭제 실행
pet_df = pet_df.drop(columns=cols_to_drop, errors='ignore')
```

```python
pet_df
```

```python
import pandas as pd

# 1. 분리 로직 함수 정의
def split_korea_addr(addr):
    if not addr or pd.isna(addr):
        return "미분류", "미분류"
    
    parts = str(addr).split()
    
    sido = parts[0] if len(parts) > 0 else "미분류"
    
    sgg = parts[1] if len(parts) > 1 else ""
    
    return sido, sgg

# 2. 새로운 컬럼 생성 (sido_nm, sgg_nm)
pet_df[['sido_nm', 'sgg_nm']] = pet_df['addr1'].apply(
    lambda x: pd.Series(split_korea_addr(x))
)

# 3. 컬럼 순서 재배치 및 불필요 컬럼 삭제.
final_cols = ['title', 'sido_nm', 'sgg_nm', 'mapx', 'mapy', 'zipcode', 'contenttypeid']
pet_df = pet_df[final_cols]
```

```python
# 1. 변환용 딕셔너리 생성
content_type_map = {
    '12': '관광지',
    '14': '문화시설',
    '15': '축제공연행사',
    '28': '레포츠',
    '32': '숙박',
    '38': '쇼핑',
    '39': '음식점'
}

# 2. 데이터 타입 맞추기 
pet_df['contenttypeid'] = pet_df['contenttypeid'].astype(str)

pet_df['content_type_nm'] = pet_df['contenttypeid'].map(content_type_map).fillna('기타')

if 'content_type_nm' in pet_df.columns:
    pet_df = pet_df.drop(columns=['contenttypeid'], errors='ignore')
    
    cols = ['title', 'content_type_nm', 'sido_nm', 'sgg_nm'] + [c for c in pet_df.columns if c not in ['title', 'content_type_nm', 'sido_nm', 'sgg_nm']]
    pet_df = pet_df[cols]
```
