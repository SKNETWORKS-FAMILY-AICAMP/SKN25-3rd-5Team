```python
import pandas as pd

# 1. 고정 컬럼 (직접 매칭이 안 될 경우를 대비해 키워드로 찾기)
fixed_keywords = [
    "성별", "연령", "반려동물 동반 여부", 
    "전반적 만족도", "재방문 의향", "타인 추천 의향"
]
fixed_cols = [c for c in tour_df.columns if any(k in c for k in fixed_keywords)]

# 2. 반복 생성 컬럼 리스트 (키워드 기반 추출)
dynamic_cols = []

# Q6-2. 여행 동반자 유형 (1~12) & Q4. 여행지 활동 (1~21)
dynamic_cols += [c for c in tour_df.columns if "여행 동반자 유형_" in c or "여행지에서의 활동_" in c]

# 3. 여행차/방문지 관련 컬럼
# 'B_여행'과 '방문지'가 동시에 들어있고, '방문 지역/숙박 지역/숙박 시설'로 끝나는 것들 추출
visit_keywords = ["방문 지역", "숙박 지역", "숙박 시설"]
visit_cols = [c for c in tour_df.columns if "B_여행" in c and "방문지" in c and any(k in c for k in visit_keywords)]

# 4. 교통 수단 관련 컬럼
transport_keywords = ["주요 이동 교통 수단"]
transport_cols = [c for c in tour_df.columns if any(k in c for k in transport_keywords)]

# 5. 모든 컬럼 합치기
target_columns = list(dict.fromkeys(fixed_cols + dynamic_cols + visit_cols + transport_cols))

# 6. 최종 데이터프레임 생성
tour_df2 = tour_df[target_columns]

# 7. 결과 확인
print(f" 추출된 컬럼 예시 (방문지 관련): {visit_cols[:3]}")
print(f" 필터링 전 컬럼 수: {tour_df.shape[1]}")
print(f" 필터링 후 컬럼 수: {tour_df2.shape[1]}")
```

```python
import pandas as pd
import numpy as np

# 1. 번호와 명칭 매핑 딕셔너리 생성
companion_map = {
    1: '배우자', 2: '자녀', 3: '손자/손녀', 4: '부모', 5: '조부모',
    6: '형제/자매', 7: '친구', 8: '연인', 9: '직장동료',
    10: '친목 단체/모임', 11: '학교 단체', 12: '기타'
}

# 2. 통합 대상 컬럼 리스트 추출 (Q6-2_1 ~ Q6-2_12)
companion_cols = [f"A.[국내] Q6-2. 여행 동반자 유형_{i}" for i in range(1, 13)]

def get_companion_type(row):
    for i, col in enumerate(companion_cols, 1):
        # 값이 1이거나, 결측치가 아니거나, 0보다 큰 경우
        if pd.notna(row[col]) and row[col] != 0:
            return companion_map.get(i, '기타')
    return '단독(없음)' # 모든 컬럼이 비어있을 경우

# 3. 새로운 통합 컬럼 생성
tour_df2['companion_type'] = tour_df2.apply(get_companion_type, axis=1)

# 4.기존의 분리된 12개 컬럼 삭제
tour_df2 = tour_df2.drop(columns=companion_cols)
```

```python
import pandas as pd

# 1. 활동 번호와 명칭 매핑 딕셔너리
activity_map = {
    1: '자연 및 풍경감상', 2: '음식관광', 3: '야외 스포츠/레포츠', 
    4: '역사 유적지 방문', 5: '테마파크/동식물원', 6: '휴식/휴양', 
    7: '온천/스파', 8: '쇼핑', 9: '문화예술/전시 관람', 
    10: '스포츠 경기관람', 11: '지역 축제/이벤트', 12: '교육/체험 프로그램', 
    13: '종교/성지순례', 14: '카지노/경마 등', 15: '시티투어', 
    16: '드라마 촬영지 방문', 17: '유흥/오락', 18: '가족/친지 방문', 
    19: '회의참가/시찰', 20: '교육/훈련/연수', 21: '기타'
}

# 2. 통합 대상 컬럼 리스트 추출 (Q4_1 ~ Q4_21)
activity_cols = [f"A.[국내] Q4. 여행지에서의 활동_{i}" for i in range(1, 22)]

# 3. 통합 함수 정의 (모든 활동을 나열하는 방식)
def get_all_activities(row):
    selected = []
    for i, col in enumerate(activity_cols, 1):
        # 값이 1(체크됨)이거나 결측치가 아닌 경우
        if pd.notna(row[col]) and row[col] != 0:
            selected.append(activity_map.get(i))
    
    return ", ".join(selected) if selected else "활동 없음"

# 4. 새로운 통합 컬럼 생성 (travel_activity)
tour_df2['travel_activity'] = tour_df2.apply(get_all_activities, axis=1)

# 5.기존의 분리된 21개 컬럼 삭제
tour_df2 = tour_df2.drop(columns=activity_cols)
```

```python
# 1. 현재 tour4의 모든 컬럼명을 가져와서 공백을 제거한 버전과 매핑
current_cols = tour4.columns.tolist()

# 2. 핵심 키워드 기반 매핑 사전
# '한글키워드': '영문명'
keyword_map = {
    '성별': 'gender',
    '연령': 'age_group',
    '1차_1번째방문지_방문지역': 'trip1_visit1_area',
    '1차_1번째방문지_숙박지역': 'trip1_stay1_area',
    '1차_1번째방문지_숙박시설': 'trip1_stay1_type',
    '1차_2번째방문지_방문지역': 'trip1_visit2_area',
    '1차_주요이동교통수단(지역간)': 'trip1_transport_intercity',
    '1차_주요이동교통수단(지역내)': 'trip1_transport_intracity_1',
    '1차_주요이동교통수단(지역내).1': 'trip1_transport_intracity_2'
}

# 3. 실제 컬럼명을 순회하며 매핑 적용
new_columns = []
for col in current_cols:
    # 현재 컬럼명에서 공백 제거
    clean_col = col.replace(" ", "")
    
    # keyword_map의 키가 clean_col 안에 포함되어 있는지 확인
    matched = False
    for k, v in keyword_map.items():
        if k in clean_col:
            new_columns.append(v)
            matched = True
            break
    
    if not matched:
        new_columns.append(col)

# 4. 컬럼명 일괄 교체
tour4.columns = new_columns

print("✅ 최종 변경된 컬럼 목록:")
print([c for c in tour4.columns if c in keyword_map.values() or c == 'with_pet'])
```

```python
# 1. tour4 데이터프레임 전체의 결측치(NaN)를 'N'으로 일괄 변경
# fillna는 기존 데이터 타입을 유지하려 하지만, 'N'이라는 문자열이 들어가면 
# 숫자 컬럼들도 자동으로 object(문자열) 타입으로 변환됩니다.
tour4 = tour4.fillna('N')

# 2. 전체 데이터프레임에 결측치가 남아있는지 확인
print(f" 전체 결측치 개수: {tour4.isnull().sum().sum()}")

# 3. 데이터 타입 변화 확인
print("\n 주요 컬럼 데이터 타입 확인:")
print(tour4.dtypes.head(10))
print("\n tour4 하위 데이터 샘플:")
print(tour4.tail())
```

```python
import pandas as pd

# 1. 새 시군구 코드 데이터 로드
code_df = pd.read_csv('./data/sigungu_code.csv', dtype=str)

# 2. 5자리 결합 코드 생성 (시도코드 2자리 + 시군구코드 3자리)
# 예: '11' + '010' -> '11010'
code_df['full_code'] = code_df['시도코드'].str.zfill(2) + code_df['시군구코드'].str.zfill(3)

# 3. 매핑용 딕셔너리 생성
sido_map = dict(zip(code_df['full_code'], code_df['시도명']))
sigungu_map = dict(zip(code_df['full_code'], code_df['시군구명']))

# 4. 변환 대상 컬럼 설정
target_cols = ['trip1_visit1_area', 'trip1_stay1_area', 'trip1_visit2_area']

for col in target_cols:
    if col in tour4.columns:
        # tour4의 데이터를 5자리 문자열로 변환
        # 'N'이거나 비어있는 값은 제외하고 처리
        tour4[col] = tour4[col].apply(
            lambda x: str(int(float(x))).zfill(5) if str(x).replace('.','').isdigit() else 'N'
        )
        
        # 새로운 컬럼명 정의 (sido, sigungu)
        base_name = col.replace('_area', '')
        sido_col = f"{base_name}_sido"
        sigungu_col = f"{base_name}_sigungu"
        
        # 매핑 수행
        tour4[sido_col] = tour4[col].map(sido_map).fillna('N')
        tour4[sigungu_col] = tour4[col].map(sigungu_map).fillna('N')

new_geo_cols = [c for c in tour4.columns if '_sido' in c or '_sigungu' in c]
print("✅ 5자리 결합 코드 기준 매핑 완료!")

# 'N'이 아닌 실제 데이터가 매핑되었는지 샘플 확인
success_sample = tour4[tour4['trip1_visit1_sido'] != 'N'][new_geo_cols].head()

if not success_sample.empty:
    print(f" 매핑 성공 컬럼: {new_geo_cols}")
    print("\n 변환 결과 샘플 (상위 5행):")
    print(success_sample)
else:
    print("\n 아직 'N'만 출력된다면, tour4['trip1_visit1_area']의 실제 값 하나만 출력해서 확인해봐야 합니다.")
    print("예: ", tour4['trip1_visit1_area'].unique()[:5])
```
